```python
temperature = 5.3
```

```python
# on arrondit le nombre au nombre à l'entier supérieur le plus proche,
temperature_celi = math.ceil(temperature)
```

```python
# nous arrondissons le nombre à l'entier inférieur le plus proche,
temperature_floor = math.floor(temperature)
```

```python
# utiliser le function round() pour arrondir la variable temperature
temperature_round = round(temperature)
```

```python
print("temperature_celi:", temperature_celi, type(temperature_celi))
print("temperature_floor:", temperature_floor, type(temperature_floor))
print("temperature_round:", temperature_round, type(temperature_round))
```

#### 3.2 Inversion de deux variables

Réalisez les étapes suivantes dans l'ordre :
 - Demande à l'utilisateur de rentrer deux valeurs et stocke-les dans deux variables différentes (par exemple `x` et `y`).
 - Effectue l'échange des valeurs entre ces deux variables.
 - Affiche la valeur des deux variables.

```python
# define 2 variables x ad y 
x = int(input("give me te value of x:"))
y = int(input("give me the value of y:"))
# print x and y
print("x:", x, "y:", y)
```

```python
# interchange the values of x and y
# Swap x and y using temp variable
temp_var = x
x = y
y = temp_var
print("new x value:", x, "new y value:", y)
```
