- Which actor is mentioned the most often?


```python
import pandas as pd

link = 'https://raw.githubusercontent.com/justmarkham/pandas-videos/master/data/imdb_1000.csv'
df = pd.read_csv(link)
df.head()
```

```python
def num_actors(a_string:str)->int:
    #num_actors = a_string.replace("[","").replace("]","").replace("u'", "").replace("'","").split(",")
    num_actors = a_string.replace("[", "").replace("]", "").replace("u'", "").replace("'", "").replace('u"', "").split(",")
    
    new_list_actors = []
    for actor in num_actors:
        actor = actor.strip()
        new_list_actors.append(actor)
    return  new_list_actors

df["actors2"] =df["actors_list"].apply(num_actors)

df_explode = df.explode("actors2")

table = pd.pivot_table(df_explode , values="title",index="actors2", aggfunc = 'count')

actor =table.index[table["title"] == table["title"].max()]
```

```python
print(f"The actor with more movies is:",actor[0])
```
