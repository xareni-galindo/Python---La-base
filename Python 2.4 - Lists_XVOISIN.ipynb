# Mission 1

Define a list, then display only the elements (and their position) that have even positions (0, 2, 4, 6...).

With `test_list = ["a", "b", "c", "d", "e"]`, the script should display:
```
a at position 0
c at position 2
e at position 4
```

````python
test_list = ["a", "b", "c", "d", "e"]

# Your code here
def elements_display(a_list:list)->str:
    for i in range(0, len(a_list), 2): #iteration using a range starting on 0, with a step of 2
        position = i # indes 
        list_element = a_list[i]  #list element
        print(f"{list_element } at posotion {position}")

elements_display(test_list)
````

### Mission 2

# Mission 3

Define a list with five numbers, then ask the user for a number. Put the number at the end of the list, remove the first element from the list, and finally display the list.

For example, with the list `test_list3 = [1, 9, 6, 15, 4]` and the input `7`, you should display:
```
[9, 6, 15, 4, 7]
```


````python
user_numer = int(input("Introduce a number:"))

def remove_append(number:int, a_list:list)->list: #defing a function to add an element on a list and erase the first element
    a_list.append(number) #adding element
    a_list.pop(0) #extracting element
    return a_list

test_list3 = [1, 9, 6, 15, 4] #test list
new_list_2 = remove_append(user_numer, test_list3)
print(f"The new list is:{new_list_2}")
````

# Mission 4

Define a list, then ask the user for a string. Then, insert the string in the list at a random position, and display the resulting list.

For example, with the list `test_list4 = ["p", "y", "t", "h", "o"]` and the input `n`, you *might* get:
```
["p", "n", "y", "t", "h", "o"]
```
... but you might also get:
```
["p", "y", "t", "n", "h", "o"]
```


````python
import random

user_string = input("Write a string") #demanding input from the user

def string_replace(string:str, a_list:list)->list: # defining a function to add an element at a random position
    index = random.randint(0,len(a_list))   #generating random number
    a_list.insert(index, string) #inserting string at the position index
    return a_list

test_list4 = ["p", "y", "t", "h", "o"]
new_list3 = string_replace(user_string, test_list4)
print(new_list3)
````

# Mission 6

Ask the user for a string, convert it to a list of characters with the `list` function, then ask for a single character.

Remove the first occurrence of the character from the list, then display how many occurrences of the character remain in the list.

For instance if you enter `"Excellent work"` then the letter `"e"`, it will display:
```
There are still 2 copies of e in the list
```

````python
users_string = input("Give me a string:")
character_ = input("Give me a single character")

#creating a function to extract the first appearance of a character and count how many times it appears again
def counting_characters(a_char:str, a_string:str)->int: 
    list_characters = list(users_string.lower()) #converting the string to lowercase
    list_characters.remove(character_) #removing character first appearence
    remaining_characters = [e for e in list_characters if e==character_] #creat a list with the remaining characters
    total_characters = len(remaining_characters) #geting the len of the list to know how many characteres remain on the list
    return total_characters

num_of_characters = counting_characters(a_char=character_, a_string=users_string)
string_to_print = f"There are still {num_of_characters} copies of {character_} in the list"
print(string_to_print)
````

# Mission 7

Define three lists containing no duplicates. Then, indicate (however you like) the two that have the most elements in common (or all three if there's a tie).

For instance, with:
```
list_a = [1, 2, 4, 8, 16, 32]
list_b = [1, 2, 3, 5, 8, 13]
list_c = [2, 3, 5, 7, 11, 13]
```
It may display:
```
Lists B and C have the most common elements!
```

````python
# defining a set of list
list_a = [1, 2, 4, 8, 16, 32]
list_b = [1, 2, 3, 5, 8, 13]
list_c = [2, 3, 5, 7, 11, 13]

#initializing 3 counters, each one representing a pair of list
counter_ab = 0
counter_ac = 0
counter_bc = 0 

# looping through the lists
for a in list_a:
    for b, c in zip(list_b, list_c):
        if b==a:
            counter_ab += 1
        elif c==a:
            counter_ac += 1
for b in list_b:
    for c in list_c:
        if c==b:
            counter_bc += 1

#comparing the counters to know which lists have more elements in common:
if (counter_ab > counter_ac) and (counter_ab > counter_bc):
    print("Lists A and B have the most common elements!")
elif  (counter_ac > counter_ab) and (counter_ac > counter_bc):
    print("Lists A and C have the most common elements!")
elif (counter_bc > counter_ab) and (counter_bc > counter_ac):
    print("Lists B and C have the most common elements!")
elif counter_ab == counter_ac == counter_bc:
    print("is a tie between the three list")
          
````
