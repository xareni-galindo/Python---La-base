```python
# Création d'une fonction merge_and_fill qui prend en paramètre
# deux DataFrame pour les joindre avec un left join sur la colonne de date

from datetime import datetime

def merge_and_fill(DataFrameMain, DataFrameOpinion):
    
  #rename the colum "date" from df_opinion to "DATE" to match df_main 
  DataFrameOpinion = DataFrameOpinion.rename(columns={"date":"DATE"})

    
  # MERGE BOTH DATAFRAMES (outer merge on date)
  NewDataFrame = pd.merge(DataFrameMain, DataFrameOpinion, how="outer", on="DATE")
  
  #change date to datetime format 
  NewDataFrame["DATE"] = pd.to_datetime(NewDataFrame["DATE"])
  
  #create 2 new columns by estracting the month and day from the date column
  NewDataFrame["MONTH"] = NewDataFrame["DATE"].dt.month
  NewDataFrame["DAY"] = NewDataFrame["DATE"].dt.day 
  
  #create a column name "FALG" to highlight the empy opinions
  NewDataFrame["FLAG"] =  NewDataFrame["OPINION"].isna()
  
  # Fill 'nan' values by "unknonw"
  NewDataFrame["OPINION"] = NewDataFrame["OPINION"].fillna("Unknown")
  
  # Fill missing values in  UV_INDEX, SUNHOUR and TOTAL_SNOW_MN
  # To fill this colums we will use a different method for each column 
  
  # WEATHER_CODE_EVENING => mode
  mode = NewDataFrame["WEATHER_CODE_EVENING"].mode()[0]
  
  NewDataFrame["WEATHER_CODE_EVENING"] = NewDataFrame["WEATHER_CODE_EVENING"].fillna(value=mode)
  
  # TOTAL_SNOW_MM => mean
  mean = NewDataFrame["TOTAL_SNOW_MM"].mean()
  NewDataFrame["TOTAL_SNOW_MM"] = NewDataFrame["TOTAL_SNOW_MM"].fillna(value=mean)
  
  # UV_INDEX => médiane
  mediane = NewDataFrame["UV_INDEX"].median()
  NewDataFrame["UV_INDEX"] = NewDataFrame["UV_INDEX"].fillna(value=mediane)
  
  # SUNHOUR => bfill
  
  NewDataFrame["SUNHOUR"] = NewDataFrame["SUNHOUR"].bfill()
  
  # Function to translate the opinion column from english to french
  def opinion_translation(row):
      
      if row == "very good":
          row = "très bien"
      elif row == "good":
          row = "bon"
      elif row =="not good not bad":
          row ="ni bon ni mauvais"
      elif row == "bad":
          row = "mauvais" 
      elif row == "very bad":
          row ="très mauvais"
      else: 
          row = "Inconnu"
           
      return row
  
  NewDataFrame["OPINION"] = NewDataFrame["OPINION"].apply(opinion_translation)
    
   
  return NewDataFrame
```

## Execution with a loop
You will execute your function in a 9 rounds loop, once per year, and agregate this in a new big DataFrame with 24 columns and lots of rows.

```python
# Chaque année entre 2010 et 2018 (inclus) a deux 2 sources : weather_main_YYYY.csv and weather_opinion_YYYY.csv
# Charger les deux datasets et joingez-les grâce à la fonction merge_and_fill
# Concaténez ce nouveau DataFrame au(x) précédent(s) pour n'obtenir qu'un long DataFrame df_weather

df_weather = pd.DataFrame()

# create a list of years to loop over    
years = list(range(2010,2019,1))

# loop for to read, merge, fill and cocatenate the data sets:    
for year in years:
    #print(year)
    # loading the main data frames
    link_main = f"https://raw.githubusercontent.com/murpi/wilddata/master/quests/weather_main_{year}.csv"
    df_main = pd.read_csv(link_main)
    #print(df_main.shape)
    
    #loading opinion dataframes
    link_opinion = f"https://raw.githubusercontent.com/murpi/wilddata/master/quests/weather_opinion_{year}.csv"
    df_opinion = pd.read_csv(link_opinion)
    #print(df_opinion.shape)
    
    #call the function merge_and_fill
    df_main_opinion_mf = merge_and_fill(df_main, df_opinion)
    
    #concatenation of de results of merge and fill function 
    df_weather = pd.concat([df_weather,df_main_opinion_mf])
   
```

```python
# Création d'une fonction clean qui :
# * crée une colonne FLAG indiquant une valeur manquante
# * convertit la colonne DATE au format datetime
# * traduit l'opinion en français

def clean(DataFrameYear):

  # ALIGN DATAFRAME SHAPE, DATATYPE
  # Colonne FLAG
  DataFrameYear["FLAG"] =  DataFrameYear["OPINION"].isna()

  # Colone DATE
  DataFrameYear["DATE"] = pd.to_datetime(DataFrameYear["DATE"])

  # Colonne OPINION
  trad_dic = {"very good":"très bien","good":"bon","not good not bad":"ni bon ni mauvais",
              "bad":"mauvais", "very bad":"très mauvais", "Unknown":"Inconnu"}

  DataFrameYear["OPINION"] = DataFrameYear["OPINION"].map(trad_dic)

  # Return new cleaned dataframe
  return DataFrameYear
```

```python
# On crée une liste de DataFrames pour les années 2019 à 2022 (inclus)
years = list(range(2019,2023,1))

# links dict initialization
links = {}

for year in range(2019, 2023):
    # Collect both DataFrames each year
    year = str(year)
    links[year] = "https://raw.githubusercontent.com/WildCodeSchool/data-training-resources/main/quests/weather" + year + ".csv"

# Create dataframes from the links to clean
dfs = {}  # Dictionary to store dataframes
for year, link in links.items():
    dfs[year] = pd.read_csv(link)
    dfs[year] = clean(dfs[year])

dfs['2019']
```

```python
# Concaténez df_weather avec df_2019 à 2022
df_19_22 = pd.DataFrame()

for df in dfs.values():
    df_19_22 = pd.concat([df_19_22,df])

df_weather = pd.concat([df_weather, df_19_22])
```
