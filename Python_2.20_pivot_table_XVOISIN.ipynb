# PIVOT TABLE

## It's like in Excel, but with Pandas

There are 2 syntaxes, choose your preferred :
- As a method : YOUR_DATAFRAME.pivot_table()
- As a function : pd.crosstab()

```python
# Example
df_weather.pivot_table(values = 'MAX_TEMPERATURE_C', index = 'MONTH', aggfunc = 'mean')
```

## Create a Pivot table with :
- Months on rows
- the maximum of Max Temperature on values
- Opinion on columns

```python
# Hop !
pd.pivot_table(df_weather, values='MAX_TEMPERATURE_C', index='MONTH', columns='OPINION')
```

## Pivot table with :
- Days on rows
- Precipitation on values
- Months on columns


```python
# Hop !
pd.pivot_table(df_weather, values='PRECIP_TOTAL_DAY_MM', index='DAY', columns='MONTH')
```

## Pivot table with :
- Months on rows
- the sum of Precipitations on values

**Then display it as a column chart.** (optionnal)

```python
# Hop !
pd.pivot_table(df_weather, values='PRECIP_TOTAL_DAY_MM', index='MONTH', aggfunc = 'sum')
```

## Pivot table with :
- Months on rows
- the average of Max Temperature on a first column of values
- the average of Sunhours on a second column of values

```python
# Hop !
table = pd.pivot_table(df_weather, values=['MAX_TEMPERATURE_C', "SUNHOUR"], index='MONTH', aggfunc = 'mean')
table
```

# Chart
Using your favorite library (MatPlotLib, Seaborn, Plotly or what you want), draw a chart from the previous pivot table with :
- Temperature in a blue column chart
- Sunhours in a red line plot

Tip : Once your pivot table created, you can help yourself with the `.reset_index()`  method.

```python
table = table.reset_index()
table
```

```python
# Your code here
import plotly.graph_objects as go
from plotly.subplots import make_subplots


fig = make_subplots(rows=1,cols=2, subplot_titles=("Temperature", "Sun hours"))

# TEMPERATURE bar chart
fig.add_trace(go.Bar(x=table["MONTH"], y=table["MAX_TEMPERATURE_C"],name="Temperature"),row=1, col=1)

fig.add_trace(go.Bar(x=table['MONTH'], y=table['SUNHOUR'],name="Sun hours"), row=1, col=2)


#fig = px.bar(table, x='MONTH', y='SUNHOUR')
fig.show()
```
