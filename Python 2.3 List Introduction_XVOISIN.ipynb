# Mission 1

For each cell, execute the code and answer the question before moving on to the next.

```python
names = ["Fredy", "Nicolas", "Alexandre"]

for i in enumerate(names):
  print(i)
```

1. What type of object is returned as output?  
Hint: You can use `type(i)`.

Answer:
La sortie de code au-dessus donne  un tuple qui est compris par l'index de chaque élément sur la liste et l'élément correspondant

```python
for i, val in enumerate(names):
    print(val)
```

2. What does `i` correspond to?  
What does `val` correspond to?  
Why is it so?

Answer: Comme on l'a indiqué précédemment, la méthode 'enumerate()' crée un tuple qui est composé de deux éléments, l'index et l'élément de la liste. Dans ce cas,  'i' est la valeur de l'index et 'val' est l'élément de la liste.

```python
for val, i in enumerate(names):
  print(val)
```

3. This time, `i` returns first names and `val` returns numbers. Why is it so?

Answer: dans la boucle 'for' on attribue les valeurs des variables 'val' et 'i' à chaque élément du tuple créé par la méthode enumerate. Où l'ordre des variables est en accord avec les valeurs des éléments du tuple. Dans ce cas, 'val' prendre le valeur des index et i le valeur des éléments sur la liste

4. What does the built in function `enumerate` do?
In what cases can it be useful?

Answer: La fonction 'enumerate()' est un outil Python qui permet de parcourir un objet itérable — tel qu'une liste, un tuple ou une chaîne de caractères — tout en ayant accès à la fois à l'index et à l'élément lui-même. Grâce à 'enumerate()' on peut avoir l'information utile de chacun des éléments sur un objet itérable au mémé temps

```python
test_list = ['Hi', 'my', 'name', 'is', "hi", "my","name", "is", "chickychicky"]

test_list[8] = "Marshall"

test_list
```

1. What does the 8 in square brackets mean?  
What action has been performed?

Answer: Le nombre '8'  représente l'index correspondant à l'élément "Marshall" qui est sur la liste 'test_list'. Dans le code ci-dessus, on a utilisé l'index '8' pour extraire la chaîne 'Marshall'.

```python
for i in range(4):
  print(test_list[i])
```

2. What does `range(4)` mean?  
Why are only the first 4 elements of `test_list` displayed?

Answer:La méthode 'range()' crée une séquence de nombres qui nous permet de créer une boucle 'for' et d'itérer sur un objet itérable quand on utilise la variable 'i' pour accéder aux éléments de cet objet. Dans ce cas, range(4) crée une séquence de nombres de 0 à 3. 

En Python, on compte en commençant par le nombre 0. Si on veut définir une séquence de nombres en utilisant la méthode range, on doit considérer son fonctionnement.Pour définir un range, on doit utiliser la notation range(start, stop, step). Start est facultatif(nombre entier indiquant la position de départ. La valeur par défaut est 0). Stop est obligatoire (nombre entier indiquant la position d'arrêt, non incluse).
Step est facultatif(nombre entier indiquant l'incrément; la valeur par défaut est 1).

Tout cela signifie que si l'on définit une suite de nombres à l'aide de `range(4)`, on obtient `range(0, 4, 1)`. La suite commence à 0, s'arrête à 3 et a un pas de 1.

Si l'on utilise cette suite pour accéder aux éléments d'une liste, on obtiendra les éléments correspondant aux nombres définis par la fonction `range(4)`, c'est-à-dire 0, 1, 2, 3.


```python
for i in range(2, 8, 2):
  print(test_list[i])
```

3. What does `range(2,8,2)` mean ?  
Why are only the elements 'name', 'hi' and 'name' displayed?

Answer: la méthode range() est définie par range(start, stop, step). Dans ce code, start=2, stop=8 et step=2. La méthode range(2, 8, 2) crée une séquence de nombres qui comprend 2, 4, 6. Dans la liste test_list = ['Hi', 'my', 'name', 'is', 'hi', 'my', 'name', 'is', 'Marshall'], les éléments correspondants aux index 2, 4 et 6 sont: name,  hi, et name.

2. What can the `.split()` method do?

Answer: La méthode split() divise une chaîne de caractères en une liste.

Answer: Dans ce cas, le caractère « à » joue le rôle de « séparateur » ; en d'autres termes, la méthode « split() » recherchera le caractère « à » et créera des sous-chaînes à partir des caractères qui le précèdent et le suivent.
