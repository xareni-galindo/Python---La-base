# Mission 2

Create a function that returns the sum of two numbers.  
The user of the function must be able to sum two numbers of his choice, when the function is executed.  
You must not use `input`.  


Question:   
What is a parameter in a function? The parameters are n1 and n2

```python
numbe_1 = float(2) # number 1 to add
number_2 = float(3) # number 2 to add

def addition(n1:float, n2:float)->float:
    suma = n1 + n2
    return suma

res = addition(n1=numbe_1, n2=number_2)
print(res)
```

# Mission 3

Create a function that returns the sum of two numbers chosen by the user of the function, in a sentence.  
For example:  
`new_sum_2(3,8)` returns `"The sum of 3 + 8 is equal to 11."`

```python
number_1 = float(input("Numner 1 = "))
number_2 = float(input("Numner 2 = "))

def print_res(n1:float, n2:float)->float:
    res = n1 + n2
    print(f" The sum of {n1} + {n2} is equal to {res}")
    
print_res(n1=number_1, n2=number_2)
```

# Mission 4

Create a function that returns 2 objects when executed:
* An object that will be the product of two numbers chosen by the user of the function.
* A word chosen by the user of the function.

```python
user_num1 = int(input("First number:"))
user_num2 = int(input("Second number"))
user_string = input("A string:") 

def double_return(n1:int, n2:int, a_string:str)->tuple:
    mutiply = n1*n2
    return mutiply, a_string
    
a_tuple = double_return(n1=user_num1, n2=user_num2, a_string=user_string)
print(f"Function return is:{a_tuple}", "its type is:", type(a_tuple))
```

Your result must be two objects that are in brackets.  
This is a tuple.

Question:  
Why is the returned object a tuple that contains two objects?  

Answer:

# Mission 5

Create a function that returns the sum of two numbers when executed.  
Also, your function must return a sentence that simply indicates if the sum of the two numbers chosen by the user is:
* less than 10
* between 10 and 50
* greater than 50  
Again, you must use `return`.

```python
def sum_(num1:float, num2:float)->float:
    sum_res = num1+num2
    if sum_res<10: 
        string_10 = f"The sum of {num1} and {num2} is less than 10"
        return sum_res, string_10
    elif 10<sum_res<50:
        string_1050 = f"The sum of {num1} and {num2} is between 10 and 50"
        return sum_res, string_1050
    elif sum_res>50:
        string_50 = f"The sum of {num1} and {num2} is bigger than 50"
        return sum_res, string_50

string_res = sum_(90,13)
print(string_res)
```
