# 1. Calcul du montant d'une remise

Un commerçant accorde une remise de 5 % pour tout achat d’un montant compris entre 100 et 500 € et 8 % au-delà.<br>
Ecrivez un programme de calcul du montant de la remise sur un achat donné.<br>
Affichez le montant de la remise.

```python
purchase_value = int(input("Purchases total amount:"))

def purchase_plus_discount(total_value):
    if total_value<100:
        final_price = total_value
        return total_value
    elif 100<=total_value<=500:
        final_price = total_value - total_value*0.05
        return final_price
    else :
        final_price = total_value - total_value*0.08
        return final_price

amount_to_pay = purchase_plus_discount(purchase_value)
print("Price to pay = ", amount_to_pay)
```

# 3. Tri de trois réels

Ecrivez un programme faisant saisir trois nombres réels x, y, z à l’utilisateur et qui les trie par ordre croissant (à la fin du déroulement du programme x ≤ y ≤ z).<br>
Affichez x, y et z.

```python
x = int(input("first number"))
y = int(input("second number"))
z = int(input("third number"))

num_list = [x, y, z]
print("the number are :", num_list)

for i in range(len(num_list)-1):
    
    v1 = num_list[i]
    v2 = num_list[i+1]
    
    if v1>v2:
        num_list[i] = v2
        num_list[i+1] = v1
        print("te sorted list is:",num_list)
    
    else:
        continue
```
