Create 2 tuples of the same length that contain only integers.

Then create a script that compares the sum of their elements and displays the tuple that has the highest total.

```python
tuple_1 =(1,2,3,4,5)
tuple_2 = (6, 7, 8, 9, 10)

def tuple_sum_comparison(a_tuple:tuple, b_tuple:tuple): #define function that compares the sum of elements in a tuple
    a_list = list(a_tuple) #cast tuples into lists
    b_list = list(b_tuple)
    
    a_sum = sum(a_list) #sum elements of lists
    b_sum = sum(b_list)

    if a_sum>b_sum: #compare sum of tuples
        print("The sum of elements in the first tuple is bigger than the sum of those on the second tuple")
        print(f"Te first tuple is {a_tuple}")
    else: 
        print("The sum of elements in the second tuple is bigger than the sum of those on the first tuple")
        print(f"Te first tuple is {b_tuple}")
        
tuple_sum_comparison(tuple_1, tuple_2 ) # calle functions
```

# Mission 5

Here's a tuple:

`my_tuple = ("data analyst", "data scientist", "data engineer", "data architect")`

Ask the user for a string, then for an integer position. Create an altered copy of `my_tuple` where the element at the given position is now the given string.

```python
user_input = input("Give me a string: ") 
postion = int(input("Give me an index: "))

def tuple_updater(a_tuple:tuple, a_string:str, an_int:int)->tuple:
    tuple_to_list = list(a_tuple)
    tuple_to_list[an_int] = a_string 
    list_to_list = tuple(tuple_to_list)
    return list_to_list

my_tuple = ("data analyst", "data scientist", "data engineer", "data architect")

my_tuple_copy = tuple_updater(a_tuple=my_tuple, a_string=user_input, an_int=postion)
print(my_tuple_copy)
```
