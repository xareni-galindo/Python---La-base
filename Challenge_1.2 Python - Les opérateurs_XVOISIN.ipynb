# 1. Calcul d'un prix

Ecrivez un programme faisant saisir à l’utilisateur un prix unitaire HT, un taux de TVA et un nombre d’articles, et qui calcule le montant TTC de l’achat.<br>
Affichez le montant TTC.

* HT: prix hors taxes
* TVA:Taxe sur la valeur ajoutée
    * Le taux normal de TVA = 20%
* TTC: prix toutes taxes comprises


```python
prix = float(input("prix HT ="))

def calcule_TTC(ht):
    TVA = 0.2
    TTC = ht + ht*TVA
    return TTC

prix_ttc = calcule_TTC(prix)
print("TTC =", prix_ttc)
```

# 2. Conversion de durées (1)

Ecrivez un programme demandant une durée en secondes à l’utilisateur, et qui la convertit en heures, minutes, secondes.<br>
Affichez le résultat.<br>
Par exemple 12334 deviendra 3 heures, 25 minutes et 34 secondes.

```python
num_seconds = int(input("Donne-moi la durée en secondes:")) #example: 12334

def seconds_converter(seconds):
    #defining the values for one minute and one hour
    divisor = 60 #60 seconds or 60 minutes
    
    #calculate how many minutes there are 
    quotient_min, remainder_secs = divmod(seconds, divisor) 
    
    #calculating how many hours there are 
    quotient_hours, remainder_min = divmod(quotient_min, divisor)
    return [quotient_hours, remainder_min, remainder_secs]

converted =  seconds_converter(num_seconds)
print(converted)
```

# 3. Conversion de durées (2)

Ecrivez un programme demandant une durée en heures, minutes, secondes à l’utilisateur et qui la convertit en secondes.<br>
Affichez le résultat.<br>
Par exemple 3 heures, 25 minutes et 34 secondes deviendra 12334 secondes.

```python
hours = int(input("Num. of hours:"))
minutes = int(input("Num. of minutes:"))
seconds = int(input("Num: of seconds:")) 

def time_converter(h, m, s):
    h_to_s = h*3600
    m_to_s = m*60
    total_seconds = h_to_s + m_to_s + s
    return total_seconds

num_seconds = time_converter(hours,minutes, seconds)
print("total of seconds:", num_seconds)
```
