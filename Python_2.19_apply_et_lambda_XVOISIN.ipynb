```python
link = 'https://raw.githubusercontent.com/chriszapp/datasets/main/books.csv'
books = pd.read_csv(link, nrows = 13)
books.head()
```

```python
# Mission 1: use apply to create a new column called 'reviews_opinion': if the ratings is between 4.5 and 5, the reviews_opinion is 'very good';
# if the rating is between 4 and 4.5, the reviews_opinion is 'good'; if the rating is between 3 and 4, the reviews_opinions will be 'neutral';
# if the rating is up to 3 the reviews_opinion is going to be 'bad'

# Add your code here
#Reviews opinion finction classification
def revs_opinions(rating:int)->str:
    if 4.5<=rating<=5:
        return 'very good'
    elif 4<=rating<4.5:
        return 'good'
    elif 3<=rating<4:
        return 'neutral'
    else:
        return 'bad'
        
books["reviews_opinion"] = books["average_rating"].apply(revs_opinions)
books.head()
```

```python
# Mission 2: using apply with a function, create a column with the number of authors per book
# (note that in the column 'authors', each name of the author is divided with the character '/')

# Add your code here
def num_authors(a_string:str)->int:
    num_authors = len(a_string.split("/"))
    return num_authors
    
books["num_authors"] = books['authors'].apply(num_authors)
books.head()
```

```python
# Mission 3: rewrite the same task as Mission 2 using the lamdba function

# Add your code here
dfBooks = books.assign(num_authors = lambda books: books['authors'].str.split("/").str.len())
```

```python
# Mission 4: create a column 'comment_ration'.
# Use the columns 'ratings_count' and 'text_reviews_count' to calculate the percentage of how many text review per reviews are per book:
# percentage = text / reviews * 100
# Use the mathematical operations between columns

# Add your code here
books['comment_ration'] = books["text_reviews_count"]/books["ratings_count"] *100
```

```python
# Mission 5: obtain the same result of Mission 4 using apply with a function.
# Try both methods to apply to the whole dataset and apply to selected columns

# Add your code here

def ratio_reviews2(reviews):
    ratio = reviews[1]/reviews[0] * 100
    return  ratio
    
books['comment_ration2'] =  books[["ratings_count","text_reviews_count"]].apply(ratio_reviews2, axis=1)
    
    
```

```python
# Mission 6: obtain the same result as Mission 5 using apply lambda
# Try both methods: apply to the whole dataset and apply to selected columns

# Add your code here
def ratio_reviews(total_reviews:int, text_reviews:int)->float:
    
    return (text_reviews/total_reviews)*100

#applying to the columns "ratings_count","text_reviews_count"]
books['comment_ration1'] = books[["ratings_count","text_reviews_count"]].apply(lambda row : ratio_reviews(row["ratings_count"],
        row["text_reviews_count"]), axis=1)    

#applying to the whole data set
books['comment_ration2'] = books.apply(lambda books : ratio_reviews(books["ratings_count"], books["text_reviews_count"]), 
                                          axis=1)    
```
